# Bronze → Silver: Wikipedia edits
Reads new rows of `bronze_wiki_edits` (written by the Eventstream Lakehouse destination) and
MERGEs them into:
* `silver_wiki_edits` – typed, de-duplicated **event history** (one row per `event_id`)
* `silver_page_latest` – **latest state**, one row per page, newest `revision_new` wins

Default trigger is `availableNow`: process everything new since the last checkpoint, then stop
(pipeline-friendly). Set `continuous = True` to keep it running every minute.

```python
continuous = False
```

```python
from wikilake import silver, tables

silver.run_stream(spark, continuous=continuous)
result = {
    "silver_edits": spark.read.table(tables.SILVER_EDITS).count(),
    "silver_pages": spark.read.table(tables.SILVER_PAGES).count(),
}
```

```python
import json
try:
    notebookutils.notebook.exit(json.dumps(result, default=str))   # value returned to the pipeline
except NameError:
    print(result)
```
