# Data-quality checks
Uniqueness, null keys, latest-state vs history consistency and freshness.
Results are appended to `ops_quality_results`; a failed check fails this activity.

```python
freshness_minutes = 120
blocking = True
```

```python
from wikilake import quality
rows = quality.run(spark, freshness_minutes=int(freshness_minutes), blocking=bool(blocking))
display(spark.createDataFrame(rows))
result = {r["check"]: r["passed"] for r in rows}
```
