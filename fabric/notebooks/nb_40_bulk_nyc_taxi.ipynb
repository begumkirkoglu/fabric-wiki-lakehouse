# Bulk load: NYC TLC yellow taxi trips (month by month)
Each month is one chunk: download → normalize to a canonical schema → write with
`replaceWhere pickup_month = 'YYYY-MM'` → record in `ops_taxi_ingest_log`.
Completed months are skipped, so a failed run simply resumes. Independent of the wiki pipeline.

```python
start_month = "2024-01"
end_month = "2024-06"
color = "yellow"
force = False
```

```python
from wikilake import nyc_taxi, tables
loaded = nyc_taxi.load(spark, start_month, end_month, color=color, force=bool(force))
display(spark.read.table(tables.TAXI_LOG).orderBy("month"))
result = {"months_loaded": [m["month"] for m in loaded], "rows": sum(m["rows"] for m in loaded)}
```

```python
import json
try:
    notebookutils.notebook.exit(json.dumps(result, default=str))   # value returned to the pipeline
except NameError:
    print(result)
```
